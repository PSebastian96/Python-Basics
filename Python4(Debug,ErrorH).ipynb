# 1. Debugging

When Python spits out an error or even when it gives you a result that is different from what you had intended, then begins the hunt for the cause of the error. Debugging is the process of finding the cause of the error in your code. When you are debugging a program, and especially if you are working on a hard bug, there are four things to try:

1. Reading: 
Examine your code, read it back to yourself, and check that it says what you meant to say
2. Running: 
Experiment by making changes and running different versions. Often if you display the right thing at the right place in the program, the problem becomes obvious, but sometimes you have to spend some time to build scaffolding
3. Ruminating: 
Take some time to think! What kind of error is it: syntax, runtime, semantic? What information can you get from the error messages, or from the output of the program? What kind of error could cause the problem you’re seeing? What did you change last, before the problem appeared
4. Retreating: 
At some point, the best thing to do is back off, undoing recent changes, until you get back to a program that works and that you understand. Then you can start rebuilding.

### 1.1. Syntax Error

These are the first errors you will make and the easiest to fix. A syntax error means that you have violated the “grammar” rules of Python. Python does its best to point right at the line and character where it noticed it was confused. The only tricky bit of syntax errors is that sometimes the mistake that needs fixing is actually earlier in the program than where Python noticed it was confused. So the line and character that Python indicates in a syntax error may just be a starting point for your investigation.

In this case, Python will raise a SyntaxError because the if statement is missing a colon (:) at the end.

```python
# Syntax Error: Missing colon at the end of the if statement
if x == 5
    print("x is 5")
```

### 1.2. Logic Error

A logic error is when your program has good syntax but there is a mistake in the order of the statements or perhaps a mistake in how the statements relate to one another. A good example of a logic error might be, “take a drink from your water bottle, put it in your backpack, walk to the library, and then put the top back on the bottle.”

In this case, the logic error is that the condition `x < 5` should be `x <= 5` to correctly identify when `x` is less than or equal to `5`.

```python
# Logic Error: Incorrect condition in the if statement
x = 5
if x < 5:
    print("x is less than 5")
elif x > 5:
    print("x is greater than 5")
else:
    print("x is equal to 5")
```

### 1.3. Semantic Error

Semantic errors occur when the code is syntactically correct and runs without raising exceptions, but it does not behave as intended due to incorrect assumptions or misunderstandings of the problem domain.

In this case, the semantic error is that the calculation of the area multiplies the length and width by 2, which is incorrect. The correct calculation should be `length * width`.

```python
# Semantic Error: Incorrect calculation of the area of a rectangle
length = 5
width = 3
area = length * width * 2  # Incorrect calculation (should be length * width)
print("Area of the rectangle:", area)
```

### 2.1. Error Types

Python `error messages` are called `exceptions`. When an error occurs, an exception is raised. All exceptions are instances of a class `BaseException`. The exceptions can be generated either by the interpreter while running the code or by functions in the code. As a developer, you can raise these exceptions to deal with errors caused by incorrect user input, for example. There are many specific exceptions in Python; for example, `ZeroDivisionError` raised when the second argument of a division or modulo operation is zero. However, if an error does not fall into one of these specific categories, a `RuntimeError` will be raised. The associated string will explain what has gone wrong.

1. `ZeroDivisionError`: Raised when division or modulo by zero is encountered.
2. `NameError`: Raised when a local or global name is not found.
3. `TypeError`: Raised when an operation or function is applied to an object of inappropriate type.
4. `ValueError`: Raised when a built-in operation or function receives an argument with the right type but an inappropriate value.
5. `IndexError`: Raised when a sequence subscript is out of range.
6. `KeyError`: Raised when a dictionary key is not found.
7. `FileNotFoundError`: Raised when a file or directory is requested but cannot be found.
8. `ImportError`: Raised when an import statement fails to find the module definition.
9. `AttributeError`: Raised when an attribute reference or assignment fails.
10. `RuntimeError`: Raised when an error is detected that doesn't fall in any of the other categories.
11. `SyntaxError`: Occurs when the Python parser encounters a syntax error, indicating an invalid structure in the code.
12. `IndentationError`: Raised when the indentation of code is incorrect, such as a mix of tabs and spaces or inconsistent indentation levels.
13. `AssertionError`: Raised when an assert statement fails, indicating that a condition specified in the assertion was `not met.
14. OverflowError`: Raised when the result of an arithmetic operation is too large to be represented within the available numeric type.
15. `FloatingPointError`: Raised when a floating-point operation fails, such as division by zero or an invalid operation.
16. `EOFError`: Raised when the input() function hits an end-of-file condition without reading any data.
17. `SystemError`: Raised when the interpreter detects an internal error in its own operations, indicating a bug in the Python interpreter itself.
18. `NotImplementedError`: Raised when a method or function has not been implemented yet, signaling that the programmer needs to provide an implementation.
19. `KeyboardInterrupt`: Raised when the user interrupts the execution of a program, typically by pressing Ctrl+C.
20. `MemoryError`: Raised when an operation runs out of memory, indicating that the system cannot allocate more memory for the operation.ration.tegories.

### 2.3. Error Handling

Error handling in Python is typically done using try-except blocks. Here's how it works:

```python
try:
    # Code that might raise an exception
    result = 10 / 0  # This will raise a ZeroDivisionError
except ZeroDivisionError:
    # Code to handle the exception
    print("Error: Division by zero!")
```

The `try` block contains the code that may raise an exception.

If an exception occurs, Python looks for an except block that matches the type of the exception.

If a matching except block is found, its code is executed.

If no matching `except` block is found, the exception propagates up to the calling code or, if unhandled, causes the program to terminate and print a traceback.

Also use else and finally blocks with try-except:

```python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Error: Division by zero!")
else:
    print("Result:", result)
finally:
    print("This will always execute.")
```

In this example:

The else block is executed if no exception occurs in the try block.
The finally block is always executed, regardless of whether an exception occurs or not. It's typically used for cleanup code (closing files, releasing resources, etc.).

# 4. Unittest

`unittest` is a built-in module in Python. It provides a framework for organizing test cases and performing unit testing. It allows developers to write test cases, test suites, and test fixtures to verify the correctness of their code. The unittest module is widely used for writing and executing unit tests in Python projects.

We have a test-driven development (TDD) example where we define a function to add two numbers and write test cases to verify its correctness using the unittest module.

```python
import unittest

# Function to add two numbers
def add_numbers(a, b):
    return a + b

# Test class for add_numbers function
class TestAddNumbers(unittest.TestCase):

    # Test case for adding positive numbers
    def test_positive_numbers(self):
        self.assertEqual(add_numbers(2, 3), 5)

    # Test case for adding negative numbers
    def test_negative_numbers(self):
        self.assertEqual(add_numbers(-2, -3), -5)

    # Test case for adding positive and negative numbers
    def test_mixed_numbers(self):
        self.assertEqual(add_numbers(2, -3), -1)

if __name__ == '__main__':
    unittest.main()
```
