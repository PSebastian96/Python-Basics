# 1. Python Syntax

### 1.1. Naming Convention

1. Python comments:

   - `#` is used before single line comment. 
   - `'''  '''` is used for multi line commments.
   - `""" """` is also used for multi line comments.

2. Functions,variable naming syntax:

    - camelCase : myVar
    - snake-case : my-var
    - underscore : my_var
    - uppercase : MY_VAR
    - PasqualCase : MyVar

### 1.2. Reserved Keywords

1. `False`: Boolean value representing false.

2. `None`: Represents the absence of a value or a null value.

3. `True`: Boolean value representing true.

4. `and`: Logical operator for conjunction (returns True if both operands are true).

5. `as`: Used in the context of an import statement to rename a module or identifier.

6. `assert`: Used for debugging purposes to test a condition and trigger an error if the condition is not true.

7. `async`: Declares an asynchronous function or a block of code.

8. `await`: Used inside an asynchronous function to pause execution until the awaited task completes.

9. `break`: Terminates the nearest enclosing loop.

10. `class`: Defines a class, a blueprint for creating objects.

11. `continue`: Skips the rest of the loop's code and continues with the next iteration.

12. `def`: Defines a function.

13. `del`: Deletes a reference to an object.

14. `elif`: Short for "else if," used in conditional statements.

15. `else`: Used in conditional statements when the if condition is false.

16. `except`: Catches and handles exceptions in exception handling.

17. `finally`: A block of code that is executed no matter what, used with exception handling.

18. `for`: Used for looping over a sequence (like a list, tuple, or string).

19. `from`: Used in import statements to specify the module to import from.

20. `global`: Declares a variable as global, meaning it can be accessed outside the current scope.

21. `if`: Conditional statement that executes code based on a specified condition.

22. `import`: Used to import modules into the current namespace.

23. `in`: Membership test, used to test if a value exists in a sequence.

24. `is`: Identity test, used to test if two objects reference the same object.

25. `lambda`: Creates an anonymous function.

26. `nonlocal`: Declares a variable as nonlocal, meaning it is not local to the current function's scope.

27. `not`: Logical operator for negation (returns True if the operand is false).

28. `or`: Logical operator for disjunction (returns True if at least one operand is true).

29. `pass`: A null operation, serves as a placeholder where syntactically some code is required.

30. `raise`: Raises an exception, often used with try and except blocks.

31. `return`: Exits a function and returns a value.

32. `try`: Used to catch and handle exceptions.

33. `while`: Used for creating loops that continue as long as a specified condition is true.

34. `with`: Simplifies resource management using a context manager.

35. `yield`: Used in generator functions to produce a series of values.

### 2.5. Logical Operators


1. Logical `NOT` Operator

- Returns TRUE if the operand is false and vice versa.

| x | NOT x | 
| :---:      |  :---:   |
| False | True |
| True  | False |

2. Logical `AND` Operator

- Returns a value/executes statement if BOTH conditions are TRUE.

| x | y | x AND Y |
| :---:      |  :---:   |   :---: |
| False | False | FALSE |
| False  | False | FALSE |
| True | False | FALSE |
| True  | True | TRUE |

3. Logical `OR` Operator

- Returns a value/executes statement if ONE OR BOTH condition(s) is/are TRUE.

| x | y | x OR Y |
| :---:      |  :---:   |   :---: |
| False | False | FALSE |
| False | True | TRUE |
| True | False | TRUE |
| True | True | TRUE |


- Logical Operator Priority:
  1. `NOT`
  2. `AND`
  3. `OR`

### 3.3. type() Function

```python
# Example variables of different data types
integer_variable = 42
float_variable = 3.14
string_variable = "Hello, World!"
list_variable = [1, 2, 3]
dictionary_variable = {'key': 'value'}
boolean_variable = True

# Using type() function to check data types
print(f' 42 is {type(integer_variable)}')       # Output: <class 'int'>
print(f'3.14 {type(float_variable)}')         # Output: <class 'float'>
print(f' "Hello, World!" is {type(string_variable)}')        # Output: <class 'str'>
print(f'[1,2,3] is {type(list_variable)}')          # Output: <class 'list'>
print(f' key: value is {type(dictionary_variable)}')    # Output: <class 'dict'>
print(f'True is {type(boolean_variable)}')       # Output: <class 'bool'>
```

### 3.4. Variables

In Python, variables are mutable, meaning their values can change during program execution. When you perform an operation like count = count + 1, you're not directly modifying the value of count. Instead, you're reassigning the variable count to reference a new value.

- Here's a breakdown of what happens:

        1. Variable Assignment: Initially, the variable count is assigned the integer value 0.
        
        2. Expression Evaluation: During each iteration of the loop, the expression count + 1 is evaluated.
        
        3. New Integer Object Creation: This expression calculates the sum of the current value of count (e.g., 0, 1, 2, etc.) and 1, resulting in a new integer object representing the incremented value.
        
        4. Variable Reassignment: The variable count is then reassigned to reference this newly created integer object.
        
        5. Loop Iteration: This process repeats for each iteration of the loop where letter == contains.

```python
count = 0

for letter in 'hello':
    if letter == 'l':
        count = count + 1

print(count)
```

In this example, the variable count starts with a `value of 0`. During each iteration of the loop, if the letter is equal to `'l'`, count is incremented by `1`. After the loop completes, count will contain the total count of occurrences of the letter `'l'` in the string `'hello'`. The key point is that count is reassigned to reference a new integer object each time it is incremented, allowing it to store the updated value.

Incrementing a string with another string is not possible in Python. Strings are immutable objects, meaning their values cannot be changed after they are created. You cannot directly modify a string by adding another string to it using the + operator.

If you attempt to increment a string with another string, you will encounter a TypeError because Python does not support adding strings to strings using the + operator.

```python
count = "0"
count = count + "1"  # This appends "1" to the string "0"
print(count)  # Output will be "01"
```

In this example, "0" is a string, and "1" is another string. When you use the + operator with strings, it performs string concatenation, meaning it joins the two strings together to create a new string."
