# 2. OS module

The `os` module in Python provides a way of using operating system-dependent functionality. It allows you to interact with the operating system, including navigating the file system, manipulating paths, working with environment variables, and executing system commands.

The `os` module provides a portable way of interacting with the underlying operating system. It abstracts away many of the differences between different operating systems, allowing you to write code that works across platforms.

### 2.1. `os` methods

**1. Working with Files and Directories:**

- os.getcwd(): Returns the current working directory.
- os.chdir(path): Changes the current working directory to the specified path.
- os.listdir(path='.'): Returns a list of the files and directories in the specified directory.
- os.mkdir(path): Creates a new directory.
- os.makedirs(path): Creates a directory and all intermediate directories.

**2. Manipulating Paths:**

- os.path.join(path1, path2, ...): Joins one or more path components intelligently.
- os.path.exists(path): Checks if a path exists.
- os.path.isdir(path): Checks if a path is a directory.
- os.path.isfile(path): Checks if a path is a regular file.
- os.path.basename(path): Returns the base name of a pathname.
- os.path.dirname(path): Returns the directory name of a pathname.

**3. Executing System Commands:**

- os.system(command): Executes the command in a subshell.
- os.spawn*(), os.exec*(), os.popen*(): Other ways of executing commands.

**4. Environment Variables:**

- os.environ: A dictionary containing the environment variables.
- os.getenv(varname): Retrieves the value of the specified environment variable.
- os.putenv(varname, value): Sets the value of the specified environment variable.

**5. Miscellaneous:**

- os.name: The name of the operating system dependent module imported.
- os.rename(src, dst): Renames a file or directory.
- os.remove(path): Removes a file.
- os.rmdir(path): Removes a directory.
- os.unlink(path): Removes a file (same as os.remove()).
- os.utime(path, times): Sets the access and modified times of a file.

### 2.2. `os` example

Below is an example demonstrating how to perform `CRUD (Create, Read, Update, Delete)` operations on text `(.txt)` and `CSV` files using the `os` module in Python:

```python
import os
import csv

# File paths
txt_file_path = "example.txt"
csv_file_path = "example.csv"

# Create or write to a text file
def create_txt_file():
    with open(txt_file_path, "w") as file:
        file.write("Hello, this is a text file.")

# Read from a text file
def read_txt_file():
    with open(txt_file_path, "r") as file:
        content = file.read()
        print("Text file content:", content)

# Append to a text file
def append_to_txt_file():
    with open(txt_file_path, "a") as file:
        file.write("\nThis is additional text appended to the file.")

# Create or write to a CSV file
def create_csv_file():
    data = [
        {"Name": "Alice", "Age": 30},
        {"Name": "Bob", "Age": 25},
        {"Name": "Charlie", "Age": 35}
    ]
    with open(csv_file_path, "w", newline="") as file:
        writer = csv.DictWriter(file, fieldnames=["Name", "Age"])
        writer.writeheader()
        writer.writerows(data)

# Read from a CSV file
def read_csv_file():
    with open(csv_file_path, "r", newline="") as file:
        reader = csv.DictReader(file)
        for row in reader:
            print(row)

# Update a CSV file
def update_csv_file():
    new_data = {"Name": "David", "Age": 40}
    with open(csv_file_path, "a", newline="") as file:
        writer = csv.DictWriter(file, fieldnames=["Name", "Age"])
        writer.writerow(new_data)

# Delete a file
def delete_file(file_path):
    if os.path.exists(file_path):
        os.remove(file_path)
        print(f"{file_path} has been deleted.")
    else:
        print(f"{file_path} does not exist.")

# Perform operations
create_txt_file()
read_txt_file()
append_to_txt_file()
read_txt_file()

create_csv_file()
read_csv_file()
update_csv_file()
read_csv_file()

# Delete files
delete_file(txt_file_path)
delete_file(csv_file_path)
```

- create_txt_file() and create_csv_file() functions create new text and CSV files, respectively, with some sample data.
- read_txt_file() and read_csv_file() functions read the contents of the text and CSV files, respectively.
- append_to_txt_file() appends additional text to the existing text file.
- update_csv_file() adds a new row to the existing CSV file.
- delete_file() function deletes a file if it exists.

Make sure to replace `"example.txt"` and `"example.csv"` with the paths to your desired text and CSV files.

### 3.3. List of Commonly Used Python Libraries:

1. NumPy:

- Purpose: NumPy is a powerful library for numerical computing in Python. It provides support for large, multi-dimensional arrays and matrices, along with a collection of mathematical functions to operate on these arrays.
- Installation: `pip install numpy`

2. Pandas:

- Purpose: Pandas is a data manipulation and analysis library. It provides data structures like DataFrame and Series, which are useful for working with structured data.
- Installation: `pip install pandas`

3. Matplotlib:

- Purpose: Matplotlib is a plotting library for Python. It allows you to create static, animated, and interactive visualizations in Python.
